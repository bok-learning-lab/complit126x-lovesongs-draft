This notebook implements the prompt chain designed by the Anne Sexton workshop group. Their chain used a **thematic extremes + synthesis** architecture — splitting Sexton's voice into two opposing poles, generating a poem at each extreme, then merging them.

```
                        ┌─────────────────┐
                        │   ★ Extreme A    │
                   ┌──▶ │ domestic/everyday │──┐
                   │    │ silly, vivid      │  │
 ┌──────────┐     │    └─────────────────┘  │   ┌──────────────────┐
 │ one-shot  │─────┤                         ├──▶│  ★ → ☽ Synthesis │──▶
 │          │     │    ┌─────────────────┐  │   │ starts in stars, │
 └──────────┘     │    │   ☽ Extreme B    │  │   │ ends on linoleum │
                   └──▶ │ darkness/eternal │──┘   └──────────────────┘
                        │ profound         │
                        └─────────────────┘
```

**Qualities the group identified:** vivid imagery, romantic darkness, visceral physicality.

**What makes this chain interesting:** Instead of asking the model to balance all of Sexton's qualities at once (which produces a flat average), the group pushed the model to two *extremes* — the domestic/everyday/silly pole and the darkness/eternal/profound pole — then asked it to synthesize. This is like exploring a poetic landscape by mapping its edges before trying to inhabit the center. Notably, the group did **not** feed in any actual Sexton poems — they worked entirely from the model's understanding.

````python
from openai import OpenAI
import json
import os

# ── API Key ──────────────────────────────────────────────────────────────────
# In Google Colab:
#   1. Click the 🔑 (Secrets) icon in the left sidebar
#   2. Add a secret named  OPENAI_API_KEY  with your key
#   3. Toggle "Notebook access" to ON, then run this cell
#
# Locally: set the OPENAI_API_KEY environment variable

try:
    from google.colab import userdata
    api_key = userdata.get('OPENAI_API_KEY')
    print("✓ Using Colab Secrets")
except (ImportError, Exception):
    api_key = os.environ.get('OPENAI_API_KEY')
    print("✓ Using environment variable")

client = OpenAI(api_key=api_key)
MODEL = "gpt-4o"

# Helper: call the model and return the text
def ask(prompt, system=None):
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.responses.create(model=MODEL, input=messages)
    return response.output_text

print(f"✓ Client ready | Model: {MODEL}")
````

---
## Step 1: One-Shot (Baseline)

Start with a simple prompt — generate a sonnet in Sexton's style. This establishes a baseline and gives us material to identify the thematic extremes.

````python
# ── Step 1: One-shot ─────────────────────────────────────────────────────────

one_shot = ask(
    "Generate a sonnet in the style of Anne Sexton."
)

print("ONE-SHOT RESULT")
print("═" * 60)
print(one_shot)
print("\n" + "═" * 60)
print("\n↓ Read this. The group found that the one-shot tries to")
print("  balance everything at once — domestic AND dark, playful")
print("  AND profound — and ends up a flat average of all of them.")
````

---
## Step 2: Identify Thematic Extremes

The group's key insight: Sexton's voice lives in the *tension* between two poles. On one side: the **domestic, everyday, silly** — kitchen tables, lipstick, the body's small comedies. On the other: the **dark, eternal, profound** — death, God, the abyss inside an ordinary afternoon.

Instead of asking the model to hit both at once, they separated them. This step identifies the poles and the key qualities that run through both: **vivid imagery, romantic darkness, visceral physicality**.

````python
# ── Step 2: Identify the extremes ────────────────────────────────────────────

extremes = ask(
    f"""Here is a one-shot sonnet generated in the style of Anne Sexton:

{one_shot}

Anne Sexton's poetry lives in the tension between two poles:

POLE A — The Domestic / Everyday / Silly:
Kitchen scenes, the body's small rituals, lipstick and laundry,
sardonic humor, the mundane made strange.

POLE B — The Dark / Eternal / Profound:
Death, madness, God, the infinite, confessional extremity,
the abyss that opens inside an ordinary room.

Running through both poles are these core qualities:
- Vivid imagery
- Romantic darkness
- Visceral physicality

Analyze the one-shot poem:
1. Which pole does it lean toward? Or does it try to do both at once?
2. How does it handle the three core qualities?
3. What gets lost when the model tries to balance everything?
4. Give specific examples of lines that are too balanced / too safe."""
)

print("THEMATIC EXTREMES ANALYSIS")
print("═" * 60)
print(extremes)
````

---
## Step 3: Extreme A — Domestic / Everyday / Silly

Now we push the model to one pole: the domestic, everyday, silly Sexton. Think of poems like "Cinderella" or "The Ballad of the Lonely Masturbator" — the Sexton who finds the grotesque and the tender in kitchen drawers and bathroom mirrors.

````python
# ── Step 3: Extreme A — domestic / everyday / silly ─────────────────────────

extreme_a = ask(
    """Write a sonnet in the style of Anne Sexton, but push it entirely
toward the DOMESTIC / EVERYDAY / SILLY pole of her work.

This is the Sexton of kitchen tables and lipstick tubes, the one who
finds dark comedy in the body's daily rituals. The poem should:

- Be grounded in a specific domestic scene (a room, an object, a routine)
- Use humor — sardonic, self-aware, even a little absurd
- Maintain vivid imagery and visceral physicality
- Feel like a confession whispered over coffee, not from a stage
- Let the darkness seep in through the details, not through
  grand declarations

Do NOT include death, God, eternity, or the infinite. Stay in the
kitchen. Stay in the body. Stay in the silly and the small.

Write only the poem."""
)

print("EXTREME A: DOMESTIC / EVERYDAY / SILLY")
print("═" * 60)
print(extreme_a)
````

---
## Step 4: Extreme B — Dark / Eternal / Profound

Now the opposite pole: the dark, eternal, profound Sexton. Think of "Wanting to Die" or "The Starry Night" — the Sexton who stares directly into the abyss and reports back with terrifying clarity.

````python
# ── Step 4: Extreme B — dark / eternal / profound ────────────────────────────

extreme_b = ask(
    """Write a sonnet in the style of Anne Sexton, but push it entirely
toward the DARK / ETERNAL / PROFOUND pole of her work.

This is the Sexton of "Wanting to Die" and "The Starry Night" — the one
who confronts death, madness, God, and the void with unflinching
directness. The poem should:

- Deal with something vast: mortality, faith, the body's betrayals,
  the pull toward oblivion
- Use vivid imagery — but make it large, cosmic, visceral
- Maintain the confessional "I" — this isn't philosophy, it's testimony
- Feel dangerous, like the speaker is saying something they shouldn't
- Let the romantic darkness be the dominant note — beauty and terror
  braided together

Do NOT include domestic details, humor, or everyday objects. Stay
in the dark. Stay in the eternal. Stay in the profound.

Write only the poem."""
)

print("EXTREME B: DARK / ETERNAL / PROFOUND")
print("═" * 60)
print(extreme_b)
````

---
## Step 5: Synthesis — Stars to Linoleum

Now the crucial step: **merge the two extremes.** The group asked for a poem that starts at Pole B (the stars, the eternal) and descends to Pole A (the linoleum, the domestic) — mirroring how Sexton herself moves between registers, often within a single stanza.

The synthesis should feel like neither extreme alone — it should feel like the tension between them, which is where Sexton's real power lives.

````python
# ── Step 5: Synthesis ────────────────────────────────────────────────────────

synthesis = ask(
    f"""Here are two sonnets in the style of Anne Sexton, each pushed
to an extreme:

--- EXTREME A: DOMESTIC / EVERYDAY / SILLY ---
{extreme_a}

--- EXTREME B: DARK / ETERNAL / PROFOUND ---
{extreme_b}

Now write a new sonnet that SYNTHESIZES these two poles. The poem should:

- **Start in the stars** — open with the eternal, the vast, the dark.
  The first lines should feel like Extreme B.
- **End on the linoleum** — descend into the domestic, the everyday,
  the body's small truths. The last lines should feel like Extreme A.
- **Move between registers** within the poem — the way Sexton herself
  jumps from God to a kitchen drawer in the same breath
- Maintain all three core qualities throughout:
  · Vivid imagery (concrete, not abstract)
  · Romantic darkness (beauty and terror together)
  · Visceral physicality (the body is always present)

The poem should feel like neither extreme alone — it should feel like
the TENSION between them. That tension is where Sexton lives.

Write only the poem."""
)

print("SYNTHESIS: STARS → LINOLEUM")
print("═" * 60)
print(synthesis)
````

---
## Compare All Versions

Now look at the full progression. The one-shot tried to be everything at once; the extremes mapped the edges; the synthesis found the tension between them.

````python
# ── Side-by-side comparison ──────────────────────────────────────────────────

print("PROGRESSION")
print("\n" + "═" * 60)
print("1. ONE-SHOT (everything at once)")
print("═" * 60)
print(one_shot)

print("\n" + "═" * 60)
print("2. EXTREME A — domestic / everyday / silly")
print("═" * 60)
print(extreme_a)

print("\n" + "═" * 60)
print("3. EXTREME B — dark / eternal / profound")
print("═" * 60)
print(extreme_b)

print("\n" + "═" * 60)
print("4. SYNTHESIS — stars → linoleum")
print("═" * 60)
print(synthesis)

print("\n" + "═" * 60)
print("\nFor your essay, consider:")
print("  → How does the one-shot differ from the synthesis?")
print("  → What did pushing to extremes reveal about the model's defaults?")
print("  → Does the synthesis actually move between registers,")
print("    or does it just alternate?")
print("  → The group didn't feed in actual poems. Would that have helped?")
print("  → What's the difference between 'balanced' and 'in tension'?")
````
